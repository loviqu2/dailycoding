```python
def greeting(name):
    message = f"Hello, {name}!"
    return message

assert greeting("Vincent") 
print("This line can run")

def test_greeting():
    assert greeting("Vincent") == "Hello, Vincent!"
    assert greeting("Alice") == "Hello, Alice!"
    print("All greeting tests passed!")

test_greeting()
```

Sometimes the correct behavior of a function is to raise an error — like set_age from Day 3, which is supposed to raise ValueError for a negative age. unittest has a specific tool for testing that: assertRaises.

```python
import unittest

def set_age(age):
    if age < 0:
        raise ValueError("Age can't be negative")
    return age

class TestSetAge(unittest.TestCase):
    def test_valid_age(self):
        self.assertEqual(set_age(25), 25)

    def test_negative_age_raises(self):
        with self.assertRaises(ValueError):
            set_age(5)

unittest.main(argv=[''], exit=False)
```
