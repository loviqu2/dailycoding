Exercise 2 — Multiple except blocks

Write a function safe_divide(a, b) that returns a / b, but handles two different possible failures with different messages:

If b is 0 → return the string "Can't divide by zero!"
If either a or b isn't a number at all (e.g. someone passes a string) → return the string "Invalid input!"

You'll need two separate except blocks under one try, each catching a different error type. Test with:

safe_divide(10, 2) → 5.0
safe_divide(10, 0) → "Can't divide by zero!"
safe_divide(10, "a") → "Invalid input!"

Hint: run 10 / 0 alone first (no handling) to find that error's name — same technique as before.

```python

def safe_divide(a, b):
    try:
        return a / b

    except TypeError:
            return ("Invalid Input")
    
    except ZeroDivisionError:
        return ("Cant Divide by zero!")

    

    
print(safe_divide(10, 2)) 
print(safe_divide(10, 0) )
print(safe_divide(10, "a"))



```

Exercise 4 — raise, creating your own errors

So far you've only caught errors Python throws automatically. But you can also deliberately trigger one yourself with raise, which is useful for enforcing rules in your own code — "this input is technically valid Python, but it violates a rule I care about."

Write a function set_age(age) that:

If age is negative, raise a ValueError with the message "Age can't be negative".
Otherwise, just return age.

```python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return age

print(set_age(25))

try:
    set_age(-5)
    
except ValueError:
    print("Invalid Age provided")
```
