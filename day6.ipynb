Generators

Generators offer a different approach: producing values one at a time, on demand, pausing in between. This matters a lot for large or even infinite sequences, where building the whole thing in memory at once would be wasteful or impossible.

```python
def make_list(n):  #create a function called make_list, insert all the numbers inside result
    result = []
    for i in range(n):
        result.append(i)
    return result

def make_generator(n): #create a function, insert the numbers one by one
    for i in range(n):
        yield i

# print(make_list(5))
# print(make_generator(5))

# gen = make_generator(3)
# print(next(gen))
# print(next(gen))
# print(next(gen))


# for value in make_generator(5):
#     print(value)


squares_list = [x**2 for x in range(5)]
squares_gen = (x**2 for x in range(5))

print(squares_list)
print(squares_gen)
```

Practical example of why generators are used.

```python
import sys

small_list = [x**2 for x in range(1000)]
small_gen = (x**2 for x in range(1000))

print(sys.getsizeof(small_list))  #sys.getsizeof tells how many bytes of memory the current code is taking up
print(sys.getsizeof(small_gen))
```

Using a generator to infinitely count

```python
def count_forever():
    n = 1 
    while True:
        yield n 
        n += 1

count = count_forever()
print(next(count))
print(next(count))
print(next(count))
```
