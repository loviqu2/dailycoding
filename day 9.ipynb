Polymorphism, if several classes have their own version of same method name, you can treat them the same way in call, where i call .speak() on any of them
and it will auto figure out which to run

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"


class Dog(Animal):  #basically means every dog is an Animal
    def speak(self):
        return f"{self.name} makes a woof" #python will check for child class first, compared to parent.

class Cat(Animal):
    def speak(self):
        return f"{self.name} makes a meow"

d = Dog("Rex")
c = Cat("R.ball")

list_dc = [d, c]  #put cat and dog into one list
for call in list_dc:
    print(call.speak())  #loop it based on the list, the output should know which to pass

# d = Dog("Rex")
# print(d.name)
# print(d.speak())
```

```python
class Engine:
    def start(self):
        return "Engine is running!"


class Car:
    def __init__(self,engine):
        self.engine = engine



# my_engine =  Engine()
# my_car = Car(my_engine)
# my_car.engine.start()

my_car = Car(Engine())
my_car.engine.start()
```

```python
class Processor:
    def run(self):
        return "Processing Data"

class Memory:
    def load(self):
        return "Loading Data"

class Computer:
    def __init__ (self, processor, memory):
        self.processor = processor
        self.memory = memory


my_comp = Computer(Processor(),Memory())  #passing two func is as usual, just need to ensure the parentheses()
print(my_comp.processor.run())
print(my_comp.memory.load())

```

Composition allows us to reuse the function without having to redo the whole function again. We use . as a way of passing onto which function we want. So in this case if i want to run processor, all i have to do is passing it through mycomp. 

If i was to use inheritance, i would need to create a new whole function

```python
class AABattery:
    def power(self):
        return "Powering with AA battery (weak, short-lasting)"

class LithiumBattery:
    def power(self):
        return "Powering with Lithium battery (strong, long-lasting)"

class Robot:
    def __init__(self, battery):
        self.battery = battery   # Robot just holds ANY battery object

    def turn_on(self):
        return self.battery.power()   # doesn't care which kind it is


#if using composition, i just have to pass on the parameter correctly, without touching function code

cheap_robot = Robot(AABattery())
print(cheap_robot.turn_on())      # "Powering with AA battery..."

strong_robot = Robot(LithiumBattery())
print(strong_robot.turn_on())     # "Powering with Lithium battery..."


class Robot(AABattery):  #if i was to use this, inheritance, then AAbattery would be placed in robot permanently.
    def turn_on(self):
        return self.power()

```
