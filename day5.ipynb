Decorators

func is a parameter name that represents "any function passed in"

```python
def shout(text):
    return text.upper() + "!"

def apply_twice(func, text):
    return func(func(text))


print(shout("Hi"))

print(apply_twice(shout, "Hello"))

```

Actual decorator

a function that carries its own private backpack of variables from wherever it was born, and can keep using them long after that original context is gone.

```python
def loud(func):
    def wrapper(**kwargs, *args):
        result = func(**kwargs, *args)
        return result 
    return wrapper


def greet(name):
    return f"Hello, {name}"

@loud
def add(a,b):
    return a + b

print(add(2, 3))


```

```python
def make_multiplier(x):
    def multiply(n):
        return x * n
    return multiply

double = make_multiplier(2)
triple = make_multiplier(5)

print(double(3))
print(triple(2))
```

the @ syntax

```python
@loud
def greet(name):
    return f"Hello, {name}"

print(greet)
print(greet("Vincent"))
```

*args, **kwargs

*args (catches any number of positional arguments into a tuple) and **kwargs (catches any number of keyword arguments into a dict)
