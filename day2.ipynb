Exercise 4 — Methods that change state

Add a new method to your Dog class called have_birthday that:

Increases self.age by 1
Returns a string like "<name> is now <age> years old!"

Then test it by calling my_dog.have_birthday() twice in a row, printing the result each time.

```python
class Dog:
    def __init__ (self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        return f"{self.name} says woof!"

    def birthday(self):
        self.age = self.age + 1
        return f"{self.name} is now {self.age} years old!"

my_dog1 = Dog("Riceball", 3)
print(my_dog1.birthday())

```
