```python
class Countdown:  #this is the countdown part, we are doing for the iteration part.
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self): 
        if self.current <= 0: #eg: value is 3, so it skips
            raise StopIteration
        value = self.current  #iterate the values to decrease each time. if until 0 or lesser, then raise value = 3
        self.current -= 1  #value becomes 2
        return value  #returns 3

    # def __next__(self):  #dunder method that auto calls the next value out of the obj
    #     if self.current <= 0: # 3
    #         raise StopIteration
    #     self.current -= 1 # if i was to do it like this, then it becomes 2 because it has minus self.current becomes 2
    #     return self.current #return whatever it is from self.current so its 2

for n in Countdown(3):
    print(n)
```
