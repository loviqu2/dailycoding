```python

# List comprehension, basic. Given nums = [3, 7, 2, 9, 4, 1, 8], write a one-line list comprehension that returns 
# only the even numbers, squared.
nums = [3, 7, 2, 9, 4, 1, 8]


for i in nums:
    if i % 2 == 0:
        print(i ** 2)

# list comprehension is basically search for nums in i, check if it is even ( i % 2 ), when they are even, then square it.

evens_squared = [i ** 2 for i in nums if i % 2 == 0]
print(evens_squared)
```

 Build a dict comprehension mapping each word to its length, but skip words with fewer than 4 letters.

search word in words, if the length of word is more/equal than 4 letters, print the word, and the length of the word

```python
words = ["apple", "kiwi", "banana", "fig"]

dict_comp = {word:len(word) for word in words if len(word) >= 4}
print(dict_comp)
```

Use a set comprehension (not set(a) & set(b)) to find the values that appear in both a and b.

```python
a = [1, 2, 2, 3, 4]
b = [3, 4, 4, 5, 6]

intersection = {x for x in a if x in b}
print(intersection)
```

Flatten this into a single list [1, 2, 3, 4, 5, 6, 7, 8, 9] using one nested comprehension — no manually-written nested for loops, no itertools.

double loop gives you into a single numbers within the matrix. 
different length of matrix does not matter

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = []


matrix2 = [[1, 2], [4, 5, 6], [7, 8, 9]]
for row in matrix2:
    for i in row:
        print(i)

for row in matrix:
    for i in row:
        flat.append(i)

print(flat)

flatten = [ i for row in matrix for i in row]
print(flatten)

print([ i for i in row for row in matrix])
```

Build a dict comprehension mapping name → "pass" if score ≥ 60 else "fail".

name,score used in loop to break the tuple into seperate type
dict comprehension is print true for the name if the score is less than 60, else print false under the loop of name, score in records.

```python
records = [("alice", 85), ("bob", 42), ("carol", 91), ("dave", 58)]

for name,score in records:
    print(name, score)

print({name: "Pass" if score >= 60 else "Fail" for name, score in records})
```

Conditional expression is put at the front for comprehension as every item is passed, it just needs to decide based on the condition
Filtering is where it decides whether the items gets passed into the result or not
