Writing my own context manager

Here's where with actually gets its power: it's not magic specific to files — it's a general pattern any class can plug into, by defining two special methods: __enter__ and __exit__. __enter__ runs when the with block starts, __exit__ runs when it ends (successfully or via a crash — same guarantee as finally).

```python
class Announcer:
    def __enter__(self):
        print("Starting...")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Finished.")
        return True

with Announcer():
    print( 1 / 0)
```
