Recursion is a function that solves a problem by calling itself, with a SMALLER version, until the problem become small enough to be solved 
directly.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1) # we need to do n * , because if n = 3, it means we need to do 3 for n and then 3-1 factorial is the remaining
    # 3 * factorial(2) = 2 * 1 

print(factorial(4))
```

Fibonacci sequence is a list where each number equals the sum of the two numbers directly before it in that same sequence. Written out: 0, 1, 1, 2, 3, 5, 8, 13... — notice 1+1=2, 1+2=3, 2+3=5, 3+5=8.

we always start with 1, 1 = 2 then 2+1, then 3 then....

```python
def fibonacci(n):
    if n == 0:
        return 0 
    elif n == 1: 
        return 1
    else:
        return fibonacci(n - 1) + fibonacci (n - 2)

print(fibonacci(6))
```

```python
data = {"a": 1, "b": {"c": 2, "d": {"e": 3, "f": 4}}, "g": 5} #total 2 recursive calls that ends with b

def sum_nested(data):
    total = 0
    for key, value in data.items():
        if isinstance(value, dict): #b onwards until the end bracket of f
            total += sum_nested(value)
        else:
            total += value  #for a and g, because the value is not a dict.
    return total

print(sum_nested(data))
```
