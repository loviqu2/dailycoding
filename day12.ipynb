lambda functions

```python
square = lambda x: x ** 2
print(square(4))

def square(x):
    return x ** 2

print(square(4))
```

```python
nums = [1, 2, 3, 4, 5]
is_even = list(map(lambda x: x % 2 == 0, nums))
print(is_even)
```

map, applies a function to every item in an iter and gives back the result

list allows me to see the output, if list is excl, then it returns normal object

```python
nums = [1, 2, 3, 4]
squared = list(map(lambda x: x ** 2, nums))
print(squared)   # [1, 4, 9, 16]
```

filter(), similar to map but returns where the func returns True

```python
nums = [1, 2, 3, 4, 5, 6, 7]
odds_only = list(filter(lambda x: x % 2 == 1, nums))
print(odds_only)
```

reduce(), collapses an iterable into a single value, by repeatedly combining two values at a time

alternative way is to use a for loop, but this method is a more compact way of doing it

```python
from functools import reduce

nums = [1, 2, 3, 4]  # maths go like 1+2, 3+3, 6+4
total = reduce(lambda a, b: a * b, nums) # ** is power, * is multiply
print(total)
```

```python
from functools import reduce

nums = [1, 2, 3, 4]  # maths go like 1+2, 3+3, 6+4
total = reduce(lambda a, b: a * b, [2, 3, 4]) # ** is power, * is multiply
print(total)
```

```python
from functools import reduce

transactions = [ 40, 160, 200, 500]

a = list(filter(lambda x: x >= 100, transactions))
print(a)

b = reduce(lambda a, b: a + b, a)
print(b)
```
