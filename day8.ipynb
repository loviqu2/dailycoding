Inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"


class Dog(Animal):  #basically means every dog is an Animal
    def speak(self):
        return f"{self.name} makes a woof" #python will check for child class first, compared to parent.

d = Dog("Rex")
print(d.name)
print(d.speak())
```

Using super

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"


class Dog(Animal):  #basically means every dog is an Animal
    def __init__(self,name, breed):
        super().__init__(name)
        self.breed = breed

    def speak(self):
        return f"{self.name} says Woof!"

d = Dog("Rex", "Golden Retriver")
print(d.name)
print(d.breed)
print(d.speak())
```

An example of why super is needed, because it is not used in Dog(animal), it would assume that energy came out of nowhere

```python
class Animal:
    def __init__(self, name):
        if name.strip() == "":
            raise ValueError("Name cannot be empty")
        self.name = name
        self.energy = 100   # every animal starts with 100 energy

class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)  # NOT calling super()

    def speak(self):
        return f"{self.name}, {self.energy}"

d =  Dog("")
print(d.energy)
print(d.speak())
```
