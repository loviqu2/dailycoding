Stack and Queue
LIFO ( Last in, First Out)
FIFO ( First in, First Out)

```python
stack = []
stack.append(10)
stack.append(20)
stack.append(30)
print(stack.pop())
```

```python
queue = []
queue.append(10)
queue.append(20)
queue.append(30)
print(queue.pop(0)) #same logic as trying to get the (-1) from an array
```

```python
from collections import deque  #deque avoids reshifting the whole item index, that can be slow when dealing with more than a few items in a list

queue = deque()
queue.append(10)
queue.append(20)
queue.append(30)
print(queue.popleft())
```

```python
stack = []
stack.append("Hanna")
stack.append("Aymen")
stack.append("Div")
stack.append("Yara")
stack.append("Vincent")

while stack:  # we need to use a while loop instead, for loop will only output some, because everytime it happens one gets pop out
    print(stack.pop())
```

```python
from collections import deque  #deque avoids reshifting the whole item index, that can be slow when dealing with more than a few items in a list

queue = deque()
queue.append("Hanna")
queue.append("Aymen")
queue.append("Div")
queue.append("Yara")
queue.append("Vincent")

while queue:
    print(queue.popleft())
```
